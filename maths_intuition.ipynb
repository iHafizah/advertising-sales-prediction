## **1.Linear Regression:**

- **Equation : Y=a+bX**

WHERE,

- Y= Varaiable which is being predicted (Target Variable)


- X= Input Variable (feature)


- a= Intercept (tells that Y is equal to what when input is Zero)


- b= Coefficient/Slope (the change in Y when Input variable increases by 1 unit)

### **Coefficient Intuition**

In multiple linear regression, a coefficient tells how much the predicted Y changes when that particular input increases by 1 unit, while the other inputs are kept constant.

Example equation:

Sales
=5+2(TV)

**Here:**

**2 = coefficient of TV**

and

5 = intercept


TV = input

Sales = predicted Sales

e.g,

| TV Value | Calculation | Predicted Sales |
|---:|---|---:|
| 0 | `5+2(0)` | 5 |
| 1 | `5+2(1)` | 7 |
| 2 | `5+2(2)` | 9 |


### **Now from project**

- *X(Input features) Columns : TV, Newspaper, Radio, TV_Radio_interaction*

- *Y(target Variable/prediced output) Column : Sales*

Linear Regression:

**Sales= a + b1(TV) +b2(Radio) +b3(Newspaper) +b4(TV_Radio_Interaction)**


EXAMPLE:

- Intercept \(a\) = 5

- TV = 100, slope \(b_1\) = 0.05

- Radio = 50, slope \(b_2\) = 0.03

- Newspaper = 20, slope \(b_3\) = 0.01

- TV × Radio = 100×50 = 5000, interaction slope \(b_4\) = 0.0002

Therefore:

$$ Sales=5+(0.05)(100)+(0.03)(50)+(0.01)(20)+(0.0002)(5000) $$ $$ \boxed{Sales=12.7} $$


**Intercept** is *The model's predicted Y when all input variables are 0.*  while each **slope/coefficient** tells *how strongly its corresponding input contributes to the predicted Sales.*

**Why the interaction term matters:** A basic equation like Sales = a + b1(TV) + b2(Radio) assumes that the effect of TV on Sales is the same regardless of the amount of Radio advertising. By adding the interaction term (TV × Radio), the model allows the effect of one variable to depend on the level of the other variable.

For example, the effect of TV on predicted Sales becomes:

$$
b_1 + b_4(Radio)
$$

So, the effect of increasing TV by 1 unit can change depending on the amount of Radio advertising.






### **OLS (Ordinary Least Squares):** How the Best Line is Found

Linear Regression doesn't guess its coefficients randomly. it uses a method called Ordinary Least Squares (OLS) to find the best possible values for a, b1, b2, b3, b4.


How it works:

- For every training row, calculate the error: **actual Sales − predicted Sales**

- Square each error (so negative and positive errors don't cancel out, and bigger mistakes are penalized more)

- Add up all the squared errors across every row

- OLS finds the exact coefficients that **make this total as small as possible**

- This is why it's called "Least Squares" , literally, the smallest sum of squared errors.




In this project: when model_1 = LinearRegression() and model_1.fit(X_train, Y_train) were run, sklearn used OLS internally to solve for the coefficients of TV, Radio, Newspaper, and TV_Radio_interaction. finding the combination that minimized total squared error across all 160 training campaigns.


### **SSE (Sum of Squared Errors)**

$$
SSE = \sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$$

**In simple words:** SSE is the total of all squared prediction errors.

OLS chooses the coefficients that make this total as small as possible.

So:

**Prediction Error → Square the Error → Add All Squared Errors → SSE → OLS Minimizes SSE**

### **Intuition Behind the OLS Formulas**

OLS wants to find the values of `a` and `b` that make the **SSE as small as possible**.

Since `a` and `b` determine the regression line, changing them changes the SSE. To find the values that give the **minimum SSE**, we use derivatives.

A derivative tells us how SSE changes when we make a small change in a particular value. At the minimum SSE, a small change in `a` or `b` cannot reduce SSE any further, so the derivative becomes **0**.

Therefore, we set:

$$
\frac{\partial SSE}{\partial a}=0
$$

$$
\frac{\partial SSE}{\partial b}=0
$$

We then solve these equations to obtain the formulas for the OLS values of `a` and `b`.

For simple Linear Regression:

$$
b=
\frac{\sum (X_i-\bar X)(y_i-\bar y)}
{\sum (X_i-\bar X)^2}
$$

$$
a=\bar y-b\bar X
$$

These formulas are therefore **derived from the goal of minimizing SSE**.


### **RMSE:** (Root Mean Squared Error)

$$RMSE = \sqrt{\frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}_i)^2}$$

**Where:**
- $y_i$ = actual Sales value for row $i$
- $\hat{y}_i$ = predicted Sales value for row $i$
- $n$ = total number of rows

**In simple words:**
RMSE takes the same squared errors used in OLS, adds them all up, divides that sum by $n$ (giving the average squared error), and then takes the square root (bringing the units back to normal Sales units instead of squared units).

**What it tells:**

RMSE measures the typical size of the model's prediction errors. Because the errors are squared before averaging, larger errors have a greater effect on RMSE. The final square root brings the result back to the original Sales units.

**In this project:**
The Linear Regression model had a Training RMSE ≈ 0.94, meaning the model's prediction errors had an RMSE of about 0.94 thousand Sales units.

### **R² (R-Squared)** : How Much Better the Model Is Than Guessing the Average

$$R^2 = 1 - \frac{\sum(y_i-\hat{y}_i)^2}{\sum(y_i-\bar{y})^2}$$

**Where:**
- $y_i$ = actual Sales value
- $\hat{y}_i$ = my model's predicted Sales value
- $\bar{y}$ = the average of all actual Sales values
- The top part = my model's actual squared errors (same as in RMSE)
- The bottom part = the error if I just guessed the average Sales value every single time, with no model at all

**In simple words:**
R² compares my real model against a baseline that just predicts the average Sales every time, no matter what. If my model's errors are much smaller than the baseline's errors, the fraction is small, so R² ends up close to 1 (good). If my model is barely better than just guessing the average, R² ends up close to 0.

**What does "variation" mean?**

Variation means how much the actual Sales values differ from their average.

For example, if Sales were:

$$
10,\ 10,\ 10,\ 10
$$

there would be almost no variation because all values are the same.

But if Sales were:

$$
5,\ 8,\ 15,\ 20
$$

there would be more variation because the values are more spread out.

Therefore, when R² is 0.97, it means the model accounts for roughly 97% of the variation in Sales values.

**In this project:**
My model's Testing R² was about 0.97, meaning it explains roughly 97% of the variation in Sales — far better than simply guessing the average Sales value for every campaign.

## **2.Ridge Regression**

$$\text{Ridge minimizes: } \sum(y_i-\hat{y}_i)^2 + \alpha\sum b_j^2$$

**Where:**
- The first part ($\sum(y_i-\hat{y}_i)^2$) = the same squared error OLS minimizes
- $b_j$ = each coefficient (TV, Radio, Newspaper, interaction)
- $\alpha$ (alpha) = a penalty strength I choose

**In simple words:**
Ridge does everything OLS does, but adds an extra penalty for having large coefficients. This discourages the model from relying too heavily on any one feature, which helps when features overlap with each other (multicollinearity). A higher alpha means a stronger penalty, shrinking coefficients more.

**In this project:**

My TV_Radio_interaction feature had moderate pairwise correlations with TV and Radio individually (correlation 0.66–0.68), showing that these predictors had some overlap. Ridge was used to see how the coefficient estimates behaved when regularization was applied. The results stayed nearly identical to plain Linear Regression across alpha value 10, suggesting that regularization did not substantially change the model in this case.
